Problema 1 - Desenvolva uma calculadora simples

```python
num1 = float(input("Digite um valor numérico: "))
num2 = float(input("Digite outro valor numérico: "))

soma = num1 + num2
diferenca = num1 - num2
produto = num1 * num2
quociente = num1 / num2
potencia = num1 ** num2
resto_divisao = num1 % num2
divisao_inteira = num1 // num2

print(f"Os resultados de cálculos dos números: {num1} e {num2}.\nSoma: {soma:.2f}.\nDiferença: {diferenca:.2f}.\nProduto: {produto:.2f}.\nQuociente: {quociente:.2f}.\nPotência: {potencia:.2f}.\nResto da divisão: {resto_divisao:.2f}.\nDivisão inteira: {divisao_inteira:.2f}.")
```
